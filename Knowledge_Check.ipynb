# What are the five limitations of Excel that make Python essential for predictive AI in offshore engineering? Provide a specific example for each limitation drawn from this module’s content.

a. Provide a specific example for each limitation drawn from this module’s content.
Ans: It can't handle the data volume.
Excel caps a sheet at 1,048,576 rows, and it slows down well before that.
Example: SCADA and metocean sensors log every second or minute for years. That is tens of millions of rows, which Python (pandas) loads and processes easily.

b. It has no built-in machine learning tools.
Ans: Excel has regression and basic stats, but no native random forests, neural networks, PCA, RFE, or physics-informed models. Python has libraries such as scikit-learn, PyTorch, and TensorFlow.
Example: Predicting corrosion rate or hydrate risk from many interacting variables needs feature engineering and model training, and Excel can't do either at scale.

c. It isn't reproducible or automatable.
Ans: Excel work is manual and click-driven, and formulas are buried in cells. It has no real version control, so it's hard to audit or repeat.
Example: A Python script reruns the same cleaning, training, and prediction steps on new jumper data every week with one command. In Excel, someone repeats the clicks by hand, and one slip can go unnoticed.

d. It struggles with complex data types.
Ans: Excel is built for neat rows and columns. Offshore data is often irregular time series, geospatial data, images, or text
Example: ROV inspection images, irregularly sampled sensor streams, and inspection report text all need Python libraries (OpenCV, pandas, NLP tools) to be usable in a model.

e. It has no proper model validation or deployment.
Ans: Predictive AI needs train/test splits, cross-validation, and hyperparameter tuning. It also needs a way to plug the trained model into live systems, and Excel supports neither properly.
Example: Without a strict train/test split you risk data leakage, where the model looks accurate but fails on new data. Python also lets you connect a model to SCADA feeds through APIs so it can flag anomalies in real time.

```python
def fahrenheit_to_celsius(temp_f: float) -> float:
    """Convert a temperature from Fahrenheit to Celsius.

    Args:
        temp_f: Temperature in degrees Fahrenheit.

    Returns:
        Temperature in degrees Celsius, rounded to 2 decimal places.

    Raises:
        TypeError: If temp_f is not a number.
        ValueError: If temp_f is below absolute zero (-459.67 F).
    """
    if isinstance(temp_f, bool) or not isinstance(temp_f, (int, float)):
        raise TypeError(f"Expected a number, got {type(temp_f).__name__}")
    if temp_f < -459.67:
        raise ValueError("Temperature cannot be below absolute zero (-459.67 F)")
    return round((temp_f - 32) * 5 / 9, 2)


# Tests
for value in [212, 39.2, "hot"]:
    try:
        print(f"{value} F = {fahrenheit_to_celsius(value)} C")
    except (TypeError, ValueError) as error:
        print(f"{value!r} -> Error: {error}")
```

Manual calculation

Barlow's formula rearranged for wall thickness: t = (P × D) / (2 × S × F × E)

Step 1: List the inputs

P = 4467 psi
D = 10.75 in
S = 65000 psi
F = 0.72 (default)
E = 1.0 (default)

Step 2: Numerator (P × D)

4467 × 10.75 = 44,670 + 3,350.25 = 48,020.25

Step 3: Denominator (2 × S × F × E)

2 × 65,000 = 130,000
130,000 × 0.72 = 93,600
93,600 × 1.0 = 93,600

Step 4: Divide

48,020.25 / 93,600 = 0.513037...

Step 5: Round to 4 decimal places

round(0.513037, 4) = 0.5130 in

```python
import math


class SubseaPipeline:
    def __init__(self, length_km: float, od_inches: float, wt_mm: float, material_grade: str):
        self.length_km = length_km
        self.od_inches = od_inches
        self.wt_mm = wt_mm
        self.material_grade = material_grade

    def internal_volume_m3(self) -> float:
        """Return the internal (bore) volume of the pipeline in cubic metres."""
        od_m = self.od_inches * 0.0254
        wt_m = self.wt_mm / 1000
        id_m = od_m - 2 * wt_m
        length_m = self.length_km * 1000
        return round(math.pi / 4 * id_m ** 2 * length_m, 2)


pipeline = SubseaPipeline(10, 10.75, 12.7, "X65")
print(pipeline.internal_volume_m3())
```

# Question 7
**Why version control is essential**

**Definition:** Version control (Git) records every change made to code: what changed, who changed it, when, and why. You can compare any two versions or roll back to an earlier one, and several people can work on the same code without overwriting each other.

**Three scenarios where Git prevents costly errors**

**1. A "small" change to a design constant goes unnoticed.**
- *Scenario:* Someone edits `design_factor` in your Barlow function from 0.72 to 0.77 while "testing," then forgets to change it back. Wall thicknesses come out thinner and nobody knows why.
- *With Git:* Every commit shows exactly which line changed, from what to what, and by whom. `git diff` finds the change in seconds, and `git revert` undoes it.
- *Cost avoided:* Under-designed pipe, or weeks of rework once the discrepancy is found.

**2. Results can't be reproduced for an audit or a failure investigation.**
- *Scenario:* Two years after installation, a regulator or client asks how the corrosion allowance for a flowline was calculated. The script has since been edited many times, and no one knows which version produced the original numbers.
- *With Git:* You check out the exact commit or tag from the date of the design report and rerun it to get the same answer.
- *Cost avoided:* Failed audits, disputed design basis, and liability exposure after an integrity incident.

**3. Two engineers overwrite each other's work.**
- *Scenario:* One engineer updates the hydrate risk model while another fixes a unit-conversion bug in the same file. Both save copies (`model_final.py`, `model_final_v2.py`, `model_FINAL_use_this.py`), and the wrong one goes into the project.
- *With Git:* Each person works on a separate branch. When merging, Git flags conflicts instead of silently overwriting, and both changes are combined deliberately.
- *Cost avoided:* A known bug reappearing in production code, or wrong inputs going to a SCADA-linked prediction tool.

# Question 8
**Calculation**

**Per day**
- Seconds per day: 24 × 60 × 60 = 86,400
- Data points: 86,400 × 5 sensors = **432,000 per day**

**Per year (365 days)**
- 432,000 × 365 = **157,680,000 data points per year**, about 158 million (158,112,000 in a leap year)

**Could Excel handle it?**

Excel allows a maximum of **1,048,576 rows per sheet**. How this compares depends on how you lay out the data:

| Layout | Rows per day | Rows per year | Days until Excel's row limit |
|---|---|---|---|
| Wide (timestamp + 5 sensor columns) | 86,400 | 31,536,000 | about 12 days |
| Long (one row per reading) | 432,000 | 157,680,000 | about 2.4 days |

**Verdict: one day, barely. One year, no.**

**Justification**
1. **Hard row limit:** A year of data is about 30 times the limit in wide format and about 150 times in long format. Excel can't even open it in a single sheet.
2. **Performance falls off well before the limit:** A sheet with hundreds of thousands of rows and formulas becomes slow, and files get very large. A single day of long-format data is already awkward to work with.
3. **Nothing is left for analysis:** Even if the data fit, cleaning it, computing rolling averages, or feeding it into an ML model would freeze or crash Excel.
4. **It's a continuous stream:** SCADA data keeps arriving, so you would have to keep splitting it across files by hand.

**Analogy:** It's like trying to store a year of subsea pipeline flow in a single notebook page. Only about 12 days of the wide layout fit before you run out of space.

**What Python does instead:** pandas can load and process tens of millions of rows in memory. For data at this scale you would use chunked reading, Parquet files, or a database, and none of that involves row limits.

The try/except pattern

Definition: try/except lets you attempt code that might fail, and catch the error instead of crashing the program. Python runs the try block first. If an error occurs, it jumps straight to the matching except block instead of stopping the script.
try:
    # code that might raise an error
except SomeErrorType:
    # what to do if that error happens

```python
def safe_divide(numerator: float, denominator: float) -> float | str:
    """Divide two numbers safely.

    Args:
        numerator: The number to be divided.
        denominator: The number to divide by.

    Returns:
        The result of the division, or an error message if the
        denominator is zero.
    """
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return "Error: Division by zero"


# Tests
print(safe_divide(48020.25, 93600))
print(safe_divide(10, 0))
```
